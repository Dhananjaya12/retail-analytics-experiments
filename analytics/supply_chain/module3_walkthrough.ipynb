## Part 1 â€” Demand analysis

**Demand** means units purchased. The SQL views aggregate demand by time, product, category, and region. A three-month or four-week rolling average smooths short-term noise. Month-over-month and week-over-week percentages describe changes; they do not predict the future.

```python
monthly = query_df("SELECT * FROM vw_demand_monthly")
weekly = query_df("SELECT * FROM vw_demand_weekly")

print(f"Monthly rows: {len(monthly):,}")
print(f"Weekly rows: {len(weekly):,}")
monthly.head()
```

```python
category_summary = (
    monthly.groupby("CATEGORY_NAME", as_index=False)
    .agg(units_sold=("UNITS_SOLD", "sum"), total_sales=("TOTAL_SALES", "sum"))
    .sort_values("units_sold", ascending=False)
)
category_summary.head(10)
```

```python
region_summary = (
    monthly.groupby("ORDER_REGION", as_index=False)
    .agg(units_sold=("UNITS_SOLD", "sum"))
    .sort_values("units_sold", ascending=False)
)
region_summary.head(10)
```

```python
category_monthly = (
    monthly.groupby(["DEMAND_MONTH", "CATEGORY_NAME"], as_index=False)
    .agg(units_sold=("UNITS_SOLD", "sum"))
)
top_categories = category_summary.head(5)["CATEGORY_NAME"]
plot_data = category_monthly[category_monthly["CATEGORY_NAME"].isin(top_categories)].copy()
plot_data["DEMAND_MONTH"] = pd.to_datetime(plot_data["DEMAND_MONTH"])

plt.figure(figsize=(13, 6))
sns.lineplot(data=plot_data, x="DEMAND_MONTH", y="units_sold", hue="CATEGORY_NAME")
plt.title("Monthly demand for the five highest-volume categories")
plt.xlabel("Month")
plt.ylabel("Units sold")
plt.show()
```

**Validated findings:** Cleats had the highest unit demand (73,734 units), and Central America had the highest regional demand (62,091 units). Units and revenue are different measures: a category can sell fewer units but generate more revenue because its products are more expensive.

## Part 2 â€” Late-delivery analysis

Late-delivery rate is `late items / all items`. We compare region, shipping mode, and category combinations. The late-risk index divides a group's rate by the overall rate: above 1 is worse than average, below 1 is better.

```python
delivery = query_df("SELECT * FROM vw_supply_chain_delivery")

numeric_columns = [
    "ORDER_ITEM_COUNT", "LATE_COUNT", "LATE_DELIVERY_RATE",
    "AVG_ACTUAL_LEAD_DAYS", "AVG_SCHEDULED_LEAD_DAYS",
    "AVG_LEAD_TIME_VARIANCE_DAYS", "LATE_RISK_INDEX"
]
delivery[numeric_columns] = delivery[numeric_columns].astype(float)

overall_late_rate = delivery["LATE_COUNT"].sum() / delivery["ORDER_ITEM_COUNT"].sum()
print(f"Overall late-delivery rate: {overall_late_rate:.1%}")
print(f"On-time fulfillment proxy: {1 - overall_late_rate:.1%}")
```

```python
mode_summary = delivery.groupby("SHIPPING_MODE", as_index=False).agg(
    items=("ORDER_ITEM_COUNT", "sum"),
    late_items=("LATE_COUNT", "sum"),
)
mode_summary["late_rate"] = mode_summary["late_items"] / mode_summary["items"]

for source, target in [
    ("AVG_ACTUAL_LEAD_DAYS", "avg_actual_days"),
    ("AVG_SCHEDULED_LEAD_DAYS", "avg_scheduled_days"),
    ("AVG_LEAD_TIME_VARIANCE_DAYS", "avg_variance_days"),
]:
    weighted = (
        delivery[source] * delivery["ORDER_ITEM_COUNT"]
    ).groupby(delivery["SHIPPING_MODE"]).sum()
    mode_summary[target] = mode_summary["SHIPPING_MODE"].map(weighted) / mode_summary["items"]

mode_summary.sort_values("late_rate", ascending=False)
```

```python
mode_plot = mode_summary.sort_values("late_rate", ascending=False)
plt.figure(figsize=(9, 5))
sns.barplot(data=mode_plot, x="SHIPPING_MODE", y="late_rate")
plt.axhline(overall_late_rate, color="red", linestyle="--", label="Overall rate")
plt.title("Late-delivery rate by shipping mode")
plt.xlabel("Shipping mode")
plt.ylabel("Late-delivery rate")
plt.legend()
plt.show()
```

**Validated finding:** faster promises were less reliable. First Class was late about 95.3% of the time and took two days against a one-day promise. Second Class was late about 76.6% of the time and took about four days against a two-day promise. Standard Class was late about 38.1% of the time and almost exactly matched its four-day promise.

## Part 3 â€” Fill-rate and lead-time proxies

The dataset has no stock levels, requested quantities, backorders, or stockout records. Therefore, we cannot calculate a true inventory fill rate. We use **on-time fulfillment rate** (`1 - late-delivery rate`) as a clearly labelled delivery-performance proxy.

Lead-time variance is `actual shipping days - scheduled shipping days`. Positive values mean slower than promised; negative values mean faster than promised. Weighted averages are used because subgroup sizes differ.

```python
high_volume = delivery[
    delivery["ORDER_ITEM_COUNT"] >= delivery["ORDER_ITEM_COUNT"].quantile(0.75)
]
high_volume.sort_values(
    ["LATE_RISK_INDEX", "ORDER_ITEM_COUNT"], ascending=False
)[[
    "ORDER_REGION", "SHIPPING_MODE", "CATEGORY_NAME", "ORDER_ITEM_COUNT",
    "LATE_DELIVERY_RATE", "LATE_RISK_INDEX", "AVG_LEAD_TIME_VARIANCE_DAYS"
]].head(15)
```

## Part 4 — Weekly category-demand forecasting

We forecast the five highest-volume categories. The last 12 observed weeks are held out as an honest test period. A 52-week seasonal-naive baseline is compared with additive Holt-Winters. The model with lower WAPE wins separately for each category.

```python
import numpy as np
from analytics.supply_chain.demand_forecasting import (
    TEST_WEEKS, FORECAST_WEEKS, make_weekly_series,
    seasonal_naive, holt_winters, mae, wape,
)

category_totals = weekly.groupby('CATEGORY_NAME')['UNITS_SOLD'].sum().sort_values(ascending=False)
top_categories = category_totals.head(5).index.tolist()
forecast_metrics = []
series_by_category = {}

for category in top_categories:
    series = make_weekly_series(weekly, category)
    series_by_category[category] = series
    train, test = series.iloc[:-TEST_WEEKS], series.iloc[-TEST_WEEKS:]
    predictions = {
        'seasonal_naive': seasonal_naive(train, TEST_WEEKS),
        'holt_winters': holt_winters(train, TEST_WEEKS),
    }
    for model_name, prediction in predictions.items():
        prediction.index = test.index
        forecast_metrics.append({
            'category': category, 'model': model_name,
            'mae_units': mae(test, prediction),
            'wape': wape(test, prediction),
        })

metrics_df = pd.DataFrame(forecast_metrics).sort_values(['category', 'wape'])
metrics_df
```

```python
winners = metrics_df.loc[
    metrics_df.groupby('category')['wape'].idxmin(),
    ['category', 'model', 'mae_units', 'wape'],
].sort_values('wape')
winners.style.format({'mae_units': '{:.1f}', 'wape': '{:.1%}'})
```

```python
example_category = winners.iloc[0]['category']
selected_model = winners.iloc[0]['model']
series = series_by_category[example_category]
future = (
    seasonal_naive(series, FORECAST_WEEKS)
    if selected_model == 'seasonal_naive'
    else holt_winters(series, FORECAST_WEEKS)
)

plt.figure(figsize=(12, 5))
history = series.iloc[-64:]
plt.plot(history.index, history.values, label='Actual')
plt.plot(future.index, future.values, '--', label=f'Forecast: {selected_model}')
plt.axvline(series.index[-1], color='gray', linestyle=':')
plt.title(f'12-week demand forecast: {example_category}')
plt.xlabel('Week')
plt.ylabel('Units sold')
plt.legend()
plt.show()
```

### Forecast interpretation

The baseline is not a throwaway model: if seasonal naive wins, the historical yearly pattern is more dependable than the fitted trend/seasonality model. If Holt-Winters wins, smoothing level, trend, and seasonality improved holdout accuracy. Either result is valid because selection is based on unseen weeks, not visual preference.

**Critical limitation:** the sharp late-2017 volume shift may be a data-collection artifact. The forecasts demonstrate a correct evaluation workflow but should not be treated as inventory recommendations without investigating that structural change.

## Conclusions and limitations

1. Demand is concentrated in a small number of categories and regions.
2. The overall late-delivery rate is high at approximately 54.8%.
3. First Class and Second Class promises appear systematically too aggressive.
4. Standard Class is slower, but promised and actual lead times are well aligned.
5. The on-time metric is a delivery proxy, not a true inventory fill rate.
6. Seasonal naive and Holt-Winters are compared using a genuine future holdout.
7. The selected 12-week forecasts demonstrate methodology, not production inventory guidance.
8. The late-2017 volume shift is the largest threat to forecast reliability.
