## 1. Why a synthetic randomized experiment?

The historical retail data was not generated by an A/B test. Existing customer segments were not randomly assigned, so comparing them cannot establish causality. We therefore randomize eligible customers 50/50 and inject effects whose true values are known: +5% order value and +5 percentage points repeat purchase. Recovering those planted effects validates the testing workflow.

```python
experiment = create_synthetic_experiment()
control = experiment[experiment['GROUP'] == 'control'].copy()
treatment = experiment[experiment['GROUP'] == 'treatment'].copy()
print(f'Eligible customers: {len(experiment):,}')
experiment.groupby('GROUP').agg(
    customers=('CUSTOMER_ID', 'count'),
    pre_aov=('PRE_EXPERIMENT_AOV', 'mean'),
    outcome_aov=('ORDER_VALUE_OUTCOME', 'mean'),
    repeat_rate=('REPEAT_PURCHASE_OUTCOME', 'mean'),
)
```

## 2. Randomization balance

Before measuring treatment outcomes, compare a pre-treatment metric. The groups should be similar because treatment assignment was random. Here, a large p-value is desirable: it means we do not detect a pre-existing AOV difference.

```python
balance = welch_t_test(control['PRE_EXPERIMENT_AOV'], treatment['PRE_EXPERIMENT_AOV'])
pd.Series(balance.__dict__)
```

**Validated result:** pre-period means were 195.87 and 197.09; p=0.199 and Cohen's d=0.028. There is no meaningful evidence of imbalance.

## 3. Continuous outcome: Welch's t-test

Average order value is continuous. Welch's t-test compares means without assuming equal variance. We report the estimate, relative lift, confidence interval, p-value, and standardized effect size.

```python
raw_result = welch_t_test(control['ORDER_VALUE_OUTCOME'], treatment['ORDER_VALUE_OUTCOME'])
pd.Series(raw_result.__dict__)
```

**Validated result:** treatment increased AOV by 10.34, or 5.12%. The 95% CI [8.58, 12.10] excludes zero, matching the planted +5% effect.

## 4. CUPED variance reduction

CUPED removes outcome variation predictable from pre-period AOV. The synthetic untreated outcome deliberately retains real pre-period behavior so CUPED can be demonstrated honestly. CUPED should reduce variance while preserving approximately the same treatment effect.

```python
cuped = apply_cuped(experiment['ORDER_VALUE_OUTCOME'], experiment['PRE_EXPERIMENT_AOV'])
experiment['ORDER_VALUE_CUPED'] = cuped.adjusted_outcome
cuped_result = welch_t_test(
    experiment.loc[experiment['GROUP'] == 'control', 'ORDER_VALUE_CUPED'],
    experiment.loc[experiment['GROUP'] == 'treatment', 'ORDER_VALUE_CUPED'],
)
pd.DataFrame({
    'metric': ['correlation', 'theta', 'variance_before', 'variance_after', 'variance_reduction'],
    'value': [cuped.correlation, cuped.theta, cuped.variance_before, cuped.variance_after, cuped.variance_reduction],
})
```

```python
comparison = pd.DataFrame({
    'method': ['Raw', 'CUPED'],
    'estimate': [raw_result.absolute_effect, cuped_result.absolute_effect],
    'ci_low': [raw_result.ci_low, cuped_result.ci_low],
    'ci_high': [raw_result.ci_high, cuped_result.ci_high],
})
comparison['ci_width'] = comparison['ci_high'] - comparison['ci_low']
comparison
```

```python
errors = np.vstack([comparison['estimate'] - comparison['ci_low'], comparison['ci_high'] - comparison['estimate']])
narrowing = 1 - comparison.loc[1, 'ci_width'] / comparison.loc[0, 'ci_width']
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
for axis, zoomed in zip(axes, [False, True]):
    axis.errorbar(comparison['estimate'], comparison['method'], xerr=errors, fmt='o', capsize=5)
    axis.axvline(0, color='gray', linestyle='--')
    axis.set_title('Zoomed CI comparison' if zoomed else 'Full scale')
    axis.set_xlabel('Estimated AOV effect (95% CI)')
    if zoomed:
        axis.set_xlim(comparison['ci_low'].min() - 0.5, comparison['ci_high'].max() + 0.5)
fig.suptitle(f'Raw vs CUPED — CI narrowed {narrowing:.1%}')
plt.show()
```

**Validated result:** pre/outcome correlation was 0.540. CUPED reduced variance by 29.2% and narrowed the confidence interval by about 16%, from width 3.52 to 2.95.

## 5. Binary outcome: repeat purchase

Repeat purchase is binary, so we use a two-proportion z-test. Absolute uplift is measured in percentage points; relative uplift compares that change with the control rate.

```python
conversion_result = two_proportion_z_test(
    int(control['REPEAT_PURCHASE_OUTCOME'].sum()), len(control),
    int(treatment['REPEAT_PURCHASE_OUTCOME'].sum()), len(treatment),
)
pd.Series(conversion_result.__dict__)
```

**Validated result:** repeat purchase increased from 45.32% to 50.53%, an absolute gain of 5.20 percentage points. The confidence interval [3.04%, 7.36%] excludes zero.

## 6. Power analysis

Power is the probability of detecting a real effect of the specified size. We compare required customers per group with the 4,095 actually available.

```python
pooled_sd = experiment['ORDER_VALUE_OUTCOME'].std(ddof=1)
continuous = continuous_power(raw_result.absolute_effect / pooled_sd, len(control))
conversion = proportion_power(conversion_result.control_rate, conversion_result.control_rate + REPEAT_PURCHASE_UPLIFT, len(control))
pd.DataFrame([
    {'metric': 'Order value', **continuous.__dict__},
    {'metric': 'Repeat purchase', **conversion.__dict__},
])
```

**Validated result:** order value required 247 customers per group and repeat purchase required 1,566. The available 4,095 per group provided approximately 100% and 99.5% achieved power.

## 7. Observational comparison — not causal

Corporate and Consumer customers were not randomly assigned. A t-test can describe their AOV association, but differences could reflect geography, product mix, or other confounding factors.

```python
observational = query_df("""
SELECT customer_segment, avg_order_value
FROM vw_customer_summary
WHERE customer_segment IN ('Consumer', 'Corporate')
""")
consumer = observational.loc[observational['CUSTOMER_SEGMENT'] == 'Consumer', 'AVG_ORDER_VALUE']
corporate = observational.loc[observational['CUSTOMER_SEGMENT'] == 'Corporate', 'AVG_ORDER_VALUE']
observational_result = welch_t_test(consumer, corporate)
pd.Series(observational_result.__dict__)
```

**Validated result:** Corporate and Consumer AOV differed by only 2.43, with p=0.601 and a confidence interval crossing zero. More importantly, even a significant observational result would not prove that segment membership caused the difference.

## Final conclusions

1. Random assignment produced balanced groups.
2. The engine recovered the planted continuous and binary effects.
3. CUPED improved precision because the pre-period covariate was correlated with the outcome.
4. The available sample was more than sufficient for both effects.
5. Randomized results support causal interpretation; existing-segment comparisons do not.
6. Synthetic effects validate the method but are not real business intervention results.
